# Sistema de Acompanhamento de Chamados

```python
from datetime import datetime, timedelta
```

```python
chamados = [
    {
        "codigo": "CH001",
        "responsavel": "Maria",
        "abertura": "15/09/2026",
        "fechamento": "17/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH002",
        "responsavel": "Carlos",
        "abertura": "16/09/2026",
        "fechamento": "22/09/2026",
        "prioridade": "Média"
    },
    {
        "codigo": "CH003",
        "responsavel": "Marina",
        "abertura": "18/09/2026",
        "fechamento": "19/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH004",
        "responsavel": "Rafael",
        "abertura": "20/09/2026",
        "fechamento": "25/09/2026",
        "prioridade": "Baixa"
    },
    {
        "codigo": "CH005",
        "responsavel": "Maria",
        "abertura": "21/09/2026",
        "fechamento": "23/09/2026",
        "prioridade": "Média"
    }
]
```

```python
for chamado in chamados:
    chamado["abertura"] = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    chamado["fechamento"] = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")

    chamado["tempo_de_atendimento_em_dias"] = (chamado["fechamento"] - chamado["abertura"]).days

    if chamado["tempo_de_atendimento_em_dias"] <= 3:
        chamado["tempo_de_atendimento"] = "Dentro do prazo"
    else:
        chamado["tempo_de_atendimento"] = "Fora do prazo"

    print(f"Código: {chamado['codigo']} | Responsável: {chamado['responsavel']} | Abertura: {chamado['abertura'].strftime('%d/%m/%Y')} | Fechamento: {chamado['fechamento'].strftime('%d/%m/%Y')} | Prioridade: {chamado['prioridade']} | Tempo de Atendimento: {chamado['tempo_de_atendimento_em_dias']} dias | Status: {chamado['tempo_de_atendimento']}")
```

```python
maiorTempo = 0
codigoMaiorTempo = ""
quantidedeChamadosDentroDoPrazo = 0
quantidadeChamadosForaDoPrazo = 0
quantidadeChamadosAltaPrioridade = 0
quantidadeChamadosMediaPrioridade = 0
quantidadeChamadosBaixaPrioridade = 0

for chamado in chamados:
    if chamado['tempo_de_atendimento'] == 'Dentro do prazo':
        quantidedeChamadosDentroDoPrazo += 1
    else:
        quantidadeChamadosForaDoPrazo += 1

    maiorTempoAtual = chamado['tempo_de_atendimento_em_dias']
    if maiorTempoAtual > maiorTempo:
        maiorTempo = maiorTempoAtual
        codigoMaiorTempo = chamado['codigo']

    if chamado['prioridade'] == 'Alta':
        quantidadeChamadosAltaPrioridade += 1
    elif chamado['prioridade'] == 'Média':
        quantidadeChamadosMediaPrioridade += 1
    else:
        quantidadeChamadosBaixaPrioridade += 1

print(f"\n quantidade de chamados: {len(chamados)}")
print(f"\n Quantidade de chamados dentro do prazo: {quantidedeChamadosDentroDoPrazo}")
print(f"\n Quantidade de chamados fora do prazo: {quantidadeChamadosForaDoPrazo}")
print(f"\n Codigo do chamado com maior tempo de atendimento: {codigoMaiorTempo} | Tempo de atendimento: {maiorTempo} dias")

print(f"\n Quantidade de chamados com alta prioridade: {quantidadeChamadosAltaPrioridade}")
print(f"\n Quantidade de chamados com média prioridade: {quantidadeChamadosMediaPrioridade}")
print(f"\n Quantidade de chamados com baixa prioridade: {quantidadeChamadosBaixaPrioridade}")
```

```python
chamados_por_prioridade = {
    "Alta": [],
    "Média": [],
    "Baixa": []
}
for chamado in chamados:
    chamados_por_prioridade[chamado["prioridade"]].append(chamado["codigo"])

print("\n Chamados por prioridade:")
print(chamados_por_prioridade)
```

```python
fora_do_prazo = []

for chamado in chamados:
    if chamado['tempo_de_atendimento'] == 'Fora do prazo':
        fora_do_prazo.append(chamado['codigo'])

print(f"\n Chamados fora do prazo: {fora_do_prazo}")
```

# Atividade Prática 2: Análise de Transações Bancárias

```python
transacoes = [
    {
        "codigo": "TR001",
        "cliente": "Maria",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "08:15",
        "processamento": 2
    },
    {
        "codigo": "TR002",
        "cliente": "Carlos",
        "tipo": "Transferência",
        "valor": 3500.00,
        "horario": "09:40",
        "processamento": 8
    },
    {
        "codigo": "TR003",
        "cliente": "Marina",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "10:25",
        "processamento": 5
    },
    {
        "codigo": "TR004",
        "cliente": "Rafael",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "11:50",
        "processamento": 3
    },
    {
        "codigo": "TR005",
        "cliente": "Beatriz",
        "tipo": "TED",
        "valor": 4200.00,
        "horario": "13:35",
        "processamento": 15
    },
    {
        "codigo": "TR006",
        "cliente": "Lucas",
        "tipo": "PIX",
        "valor": 2100.00,
        "horario": "14:20",
        "processamento": 2
    },
    {
        "codigo": "TR007",
        "cliente": "Fernanda",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "15:45",
        "processamento": 6
    }
]
```

```python
for ordem, transacao in enumerate(transacoes, start=1):
    transacao["horario_inicio"] = datetime.strptime(transacao["horario"], "%H:%M")
    transacao["horario_conclusao"] = transacao["horario_inicio"] + timedelta(minutes=transacao["processamento"])
        
    print(f"Ordem: {ordem} | Código: {transacao['codigo']} | Cliente: {transacao['cliente']} | "
          f"Tipo: {transacao['tipo']} | Valor: R${transacao['valor']:.2f} | "
          f"Horário de Início: {transacao['horario_inicio'].strftime('%H:%M')} | "
          f"Horário de Conclusão: {transacao['horario_conclusao'].strftime('%H:%M')}")
```

```python
import statistics
from collections import Counter
```

```python
valores_transacoes = [t['valor'] for t in transacoes]
media = statistics.mean(valores_transacoes)
moda = statistics.mode(valores_transacoes)
mediana = statistics.median(valores_transacoes)
variancia = statistics.variance(valores_transacoes)
desvio_padrao = statistics.stdev(valores_transacoes)

print(f"\n Média das transações: R${media:.2f}")
print(f"Moda das transações: R${moda:.2f}")
print(f"Mediana das transações: R${mediana:.2f}")
print(f"Variancia das transações: R${variancia:.2f}")
print(f"Desvio Padrão das transações: R${desvio_padrao:.2f}")
print(f"Menor valor de transação: R${min(valores_transacoes):.2f}")
print(f"Maior valor de transação: R${max(valores_transacoes):.2f}")
```

```python
for transacao in transacoes:
    if transacao['valor'] > media:
        transacao['status_media'] = "Acima da média"
    elif transacao['valor'] < media:
        transacao['status_media'] = "Abaixo da média"
    else:
        transacao['status_media'] = "Na média"
```

```python
for transacao in enumerate(transacoes, start=1):
    print(f"Ordem: {transacao[0]} | Código: {transacao[1]['codigo']} | Cliente: {transacao[1]['cliente']} | Valor: R${transacao[1]['valor']:.2f} | Status em relação à média: {transacao[1]['status_media']}")
```

```python
for transacao in transacoes:
    if transacao['valor'] == max(valores_transacoes):
        codigoMaiorTransacao = transacao['codigo']
    if transacao['valor'] == min(valores_transacoes):
        codigoMenorTransacao = transacao['codigo']

quantidade_transacoes_acima_da_media = 0
quantidade_transacoes_abaixo_da_media = 0

for transacao in transacoes:
    if transacao['valor'] > media:
        quantidade_transacoes_acima_da_media += 1
    elif transacao['valor'] < media:
        quantidade_transacoes_abaixo_da_media += 1


print(f"Quantidade de transações: {len(transacoes)}")
print(f"Valor total das transações: R${sum(valores_transacoes):.2f}")
print(f"Valor médio das transações: R${media:.2f}")
print(f"Código da transação com maior valor: {codigoMaiorTransacao}")
print(f"Código da transação com menor valor: {codigoMenorTransacao}")
print(f"Quantidade de transações acima da média: {quantidade_transacoes_acima_da_media}")
print(f"Quantidade de transações abaixo da média: {quantidade_transacoes_abaixo_da_media}")
```
