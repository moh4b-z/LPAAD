```python


def corTexto(texto, cor, estilo='nada'):
    cores = {
        'vermelho': '\033[31m',
        'verde': '\033[32m',
        'amarelo': '\033[33m',
        'azul': '\033[34m',
        'magenta': '\033[95m',
        'ciano': '\033[36m',
        'branco': '\033[97m'
    }
    estilos = {
        'nada': '\033[0m',
        'negrito': '\033[1m',
        'sublinhado': '\033[4m',
    }

    return f"{estilos[estilo]}{cores[cor]}{texto}\033[0m"

print(f"{corTexto("Olá, mundo!", "vermelho", "negrito")} e {corTexto("Olá, mundo!", "azul", "sublinhado")}")
```

```python
import time

def tempoTexte():
    print(corTexto("Iniciando contagem regressiva...", "amarelo", "negrito"))
    time.sleep(1)
    print(corTexto("3", "vermelho", "negrito"))

tempoTexte()
```

```python
def mediana(lista):
    lista_ordenada = sorted(lista)
    n = len(lista_ordenada)
    meio = n // 2

    if n % 2 == 0:
        return (lista_ordenada[meio - 1] + lista_ordenada[meio]) / 2
    else:
        return lista_ordenada[meio]

conjuntoA = [23,14,31,17,20]
conjuntoB = [75,1,4,3,19,50,50,64,31]

print(mediana(conjuntoA))
print(mediana(conjuntoB))
```

```python
lista = [90, 0, 0, 30, 40]
def moda(lista):

    moda = 1
    for valor in lista:
        print(f"{lista.count(valor)} -- {moda}")
        if lista.count(valor) > moda:
            moda = valor

    return moda

print(moda(lista))
print(lista)
```
