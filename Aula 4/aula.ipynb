```python
estoque = [18, 5 ,0 , 32,4,7,0,14,3]

totalQuantidade = len(estoque)
esgotados = 0
estoqueCritico = 0

for produto in estoque:
    if produto == 0:
        esgotados += 1
    elif produto <= 5:
        estoqueCritico += 1
por = (esgotados + estoqueCritico) / totalQuantidade * 100
print(f"Quantidades de esgotados: {esgotados}, em quantidade critica: {estoqueCritico}. Porcentagem do estoque que está precisando de reposição: {por:.2f}%")
```

```python
unidades = [
    [120, 135, 128, 141],
    [98, 105, 110, 103],
    [150, 145, 160, 155],
    [116, 161, 108, 121]
]
maiorValor = 0
abaixoDe110 = []
unidadeMaior = [0, 0]

for i, unidade in enumerate(unidades):
    for semana in unidade:
        if semana < 110:
            abaixoDe110.append(semana)
        if semana > maiorValor:
            maiorValor = semana
        
    total_da_unidade = sum(unidade)
    unidade.append(total_da_unidade)

    if unidade[4] > unidadeMaior[0]:
        unidadeMaior[0] = unidade[4]
        unidadeMaior[1] = i + 1
    


for i in range(len(unidades)):
    print(f"Unidade {i + 1} fez um total de {unidades[i][4]} de atendimento")

print(f"Maior quantidade de atendimento foi de {maiorValor}")
print(f"Maior quantidade de atendimento foi da unidade {unidadeMaior[1]}")
print(f"Teve {len(abaixoDe110)} semanas com menos que 110 atendimentos")
```
