```python
from datetime import datetime

agora = datetime.now()
print(agora)
```

```python
data_formatada = agora.strftime("%d/%m/%Y %H:%M:%S")
print(data_formatada)
```

```python
data_pedido = "12/09/2026"
data_entrega = "18/09/2026"

print(type(data_pedido))
print(type(data_entrega))

data_pedido_formatada = datetime.strptime(data_pedido, "%d/%m/%Y")
data_entrega_formatada = datetime.strptime(data_entrega, "%d/%m/%Y")

print(type(data_pedido_formatada))
print(type(data_entrega_formatada))
```

```python
diferencia_dias = data_entrega_formatada - data_pedido_formatada
print(diferencia_dias.days)
```

```python
if diferencia_dias.days > 5:
    print("O tempo da entrega foi maior que 5 dias.")
else:
    print("O tempo da entrega foi menor ou igual a 5 dias.")
```

# enumerate()

```python
vendedores = [
    {"nome": "Ana", "vendas": 12500},
    {"nome": "Carlos", "vendas": 9800},
    {"nome": "Marina", "vendas": 15700},
    {"nome": "Rafael", "vendas": 11300},
]

for ordem, vendedor in enumerate(vendedores, start=1):
    meta = ""
    if vendedor['vendas'] >= 10000:
        meta = "Meta atingida"
    else:
        meta = "Meta não atingida"
    print(f"{ordem}º - {vendedor['nome']} - Vendas: R${vendedor['vendas']} - {meta}")
```
