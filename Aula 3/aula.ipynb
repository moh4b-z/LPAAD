```python
tempos = [32, 48, 27, 55, 41, 29]
maximo = 40
somaTempo = 0
contador = 1
erros = 0


for tempo in tempos:
    somaTempo += tempo
    condicao = "No tempo"
    if tempo > maximo:
        condicao = "Atrasada"
        erros += 1
    print("Indice da entrega ", contador, "tempo: ", tempo, "a condição:", condicao)
    contador += 1

total = len(tempos)
por =   erros / total * 100
print(total)
print(somaTempo / total)
print(erros)
print(por, "%")
```

```python
produces = []

pergunta = input("Para iniciar as perguntas de valores digite qualquer tecla e confirme, se não escreva (NÃO)")
while pergunta != "NÃO":
    valor = int(input("Informe o valor entre 300 a 800:"))
    if valor >= 300 and valor <= 800:
        produces.append(valor)
    else:
        print("O valor precisa está entre 300 a 800")
    pergunta = input("Para iniciar as perguntas de valores digite qualquer tecla e confirme, se não escreva (NÃO)")

if len(produces) > 0:
    total = sum(produces)
    turnos = len(produces)
    acima = 0
    turno = 1
    for produto in produces:
        if produto >= 500:
            print("Turno", turno, "° meta de 500 peças atingida com", produto, "de peças feitas.")
            acima += 1
        elif produto >= 400:
            print("Turno", turno, "° abaixo da meta de 500 peças com", produto, "de peças feitas.")
        else:
            print("Turno", turno, "° crítico, abaixo de 400 peças com", produto, "de peças feitas.")
        turno += 1

    print("Produção total:", total)
    print("Turnos:", turnos)
    print("A média de peças foram de",  total / turnos)
    print("Quantidade de turnos que atigiram a meta:", acima)
    print("Porcentagem que atigiram a meta:", acima / turnos * 100, "%")
```
