# Problema 1

```python
import statistics
from collections import Counter
```

```python
def calcular_modalidade(dados):
    if not dados:
        return {"valores": [], "modalidade": "amodal"}
    
    contagem = Counter(dados)
    frequencia_maxima = max(contagem.values())
    
    if frequencia_maxima == 1:
        return {"valores": sorted(list(contagem.keys())), "modalidade": "amodal"}
    
    modas = [item for item, freq in contagem.items() if freq == frequencia_maxima]
    modas.sort()
    
    quantidade = len(modas)
    if quantidade == 1:
        modalidade = "unimodal"
    elif quantidade == 2:
        modalidade = "bimodal"
    else:
        modalidade = "multimodal"
        
    return {"valores": modas, "modalidade": modalidade}
```

```python
def calcular_quartil(dados):
    if not dados:
        return {"Q1": None, "Q2": None, "Q3": None}
    
    dados_ordenados = sorted(dados)
    n = len(dados_ordenados)
    
    Q2 = statistics.median(dados_ordenados)
    
    if n % 2 == 0:
        lower_half = dados_ordenados[:n // 2]
        upper_half = dados_ordenados[n // 2:]
    else:
        lower_half = dados_ordenados[:n // 2]
        upper_half = dados_ordenados[n // 2 + 1:]
    
    Q1 = statistics.median(lower_half)
    Q3 = statistics.median(upper_half)

    if Q1 is None or Q3 is None:
            return None
    
    return {"Q1": Q1, "Q2": Q2, "Q3": Q3, "IQR": Q3 - Q1}
```

```python
def estatísticasGerais(lista):
    media = statistics.mean(lista)
    mediana = statistics.median(lista)
    moda = calcular_modalidade(lista)
    variancia_amostral = statistics.variance(lista)
    desvio_padrao_amostral = statistics.stdev(lista)
    variancia_populacional = statistics.pvariance(lista)
    desvio_padrao_populacional = statistics.pstdev(lista)
    maximo = max(lista)
    minimo = min(lista)
    amplitude = maximo - minimo
    coeficiente_variacao_amostral = (desvio_padrao_amostral / media) * 100 
    coeficiente_variacao_populacional = (desvio_padrao_populacional / media) * 100 
    quartis = calcular_quartil(lista)

        
    valores = {
        "Media": media,
        "Mediana": mediana,
        "Moda": moda,
        "Variancia_Amostral": variancia_amostral,
        "DesvioPadrao_Amostral": desvio_padrao_amostral,
        "Variancia_Populacional": variancia_populacional,
        "DesvioPadrao_Populacional": desvio_padrao_populacional,
        "Maximo": maximo,
        "Minimo": minimo,
        "Amplitude": amplitude,
        "CoeficienteVariacao_Amostral": coeficiente_variacao_amostral,
        "CoeficienteVariacao_Populacional": coeficiente_variacao_populacional,
        "Quartis": quartis
    }
    return valores
```

```python
equipeA = [42, 38, 45, 41, 39, 44, 38]
equipeB = [30, 52, 35, 48, 29, 55, 38]

print("Estatísticas da equipe A:")
print(estatísticasGerais(equipeA))
print("\nEstatísticas da equipe B:")
print(estatísticasGerais(equipeB))
```
