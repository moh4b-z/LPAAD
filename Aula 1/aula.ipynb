```python
temperaturaDas8 = 72.3
temperaturaDas9 = 73.1
temperaturaDas10 = 74.0
temperaturaDas11 = 72.8
temperaturaDas12 = 71.5
temperaturaDas13 = 75.2
temperaturaDas14 = 73.6
temperaturaDas15 = 74.4

mediaTemperatura = (temperaturaDas8 + temperaturaDas9 + temperaturaDas10 + temperaturaDas11 + temperaturaDas12 + temperaturaDas13 + temperaturaDas14 + temperaturaDas15) / 8
print("A média das temperaturas é: ", mediaTemperatura)
print(f"A maior temperatura foi a {temperaturaDas13}°C e a menor foi a {temperaturaDas12}°C")
print(f"A amplitude térmica foi de {temperaturaDas13 - temperaturaDas12}°C")
print(f"Diferença da media para a temperatura ideal de 73°C é de {mediaTemperatura - 73}°C")
```

```python
producaoDeJan = 1140
pecasDefeituosasDeJan = 25
producaoDeFev = 1210
pecasDefeituosasDeFev = 22
producaoDeMar = 1250
pecasDefeituosasDeMar = 16

taxaDeRejeicaoDeJan = (pecasDefeituosasDeJan / producaoDeJan) * 100
taxaDeRejeicaoDeFev = (pecasDefeituosasDeFev / producaoDeFev) * 100
taxaDeRejeicaoDeMar = (pecasDefeituosasDeMar / producaoDeMar) * 100

taxaDeAceitacaoDeJan = 100 - taxaDeRejeicaoDeJan
taxaDeAceitacaoDeFev = 100 - taxaDeRejeicaoDeFev
taxaDeAceitacaoDeMar = 100 - taxaDeRejeicaoDeMar

mediaDeRejeicao = (taxaDeRejeicaoDeJan + taxaDeRejeicaoDeFev + taxaDeRejeicaoDeMar) / 3

print(f"A taxa de rejeição de Janeiro foi de {taxaDeRejeicaoDeJan:.2f}% e a taxa de aceitação foi de {taxaDeAceitacaoDeJan:.2f}%")
print(f"A taxa de rejeição de Fevereiro foi de {taxaDeRejeicaoDeFev:.2f}% e a taxa de aceitação foi de {taxaDeAceitacaoDeFev:.2f}%")
print(f"A taxa de rejeição de Março foi de {taxaDeRejeicaoDeMar:.2f}% e a taxa de aceitação foi de {taxaDeAceitacaoDeMar:.2f}%")
print(f"A média de rejeição dos três meses foi de {mediaDeRejeicao:.2f}%")
```
