```python
dias = [18500, 21300, 17800, 24600, 27400]

faturamento_total = sum(dias)
maximo_faturamento = max(dias)
minimo_faturamento = min(dias)
media_faturamento = faturamento_total / len(dias)

print("Quantidade de dias:", len(dias))
print("Faturamento total:", faturamento_total)
print("Máximo faturamento:", maximo_faturamento)
print("Mínimo faturamento:", minimo_faturamento)
print("Média faturamento:", media_faturamento)

if media_faturamento > 22000:
    print("Bom desempenho")
elif media_faturamento < 19000:
    print("Desempenho critico")
else:
    print("Desempenho regular")
```

```python
velocidade = int(input("Digite a velocidade do carro: "))
limite_velocidade = 70
if velocidade > limite_velocidade:
    diferencia = velocidade - limite_velocidade
    valor_multa = diferencia * 9
    porcentagem_excesso = (diferencia / limite_velocidade) * 100
    print("Você foi multado")
    print("O valor da multa é:", valor_multa)
    print("Você excedeu o limite de velocidade em", porcentagem_excesso, "%")
    if porcentagem_excesso >= 50:
        print("Ficara um ano sem carteira de habilitação")
else:
    print("Você está dentro do limite de velocidade")
```

```python
producao = [485, 510, 472, 498, 455, 520]

producao_total = sum(producao)
turnos = len(producao)
media_producao = producao_total / turnos
maior_producao = max(producao)
menor_producao = min(producao)

print("Produção total:", producao_total)
print("Número de turnos:", turnos)
print("Média de produção:", media_producao)
print("Maior produção:", maior_producao)
print("Menor produção:", menor_producao)

if media_producao > 500:
    print("Meta atingida")
elif media_producao < 480:
    print("Abaixo da meta")
else:
    print("Atenção")

if menor_producao < 460:
    print("Alerta operacional")
```

```python
janeiro = [85000, 46000,18000]
fevereiro = [92000, 51000, 19000]
marco = [78000, 48000, 20000]
abril = [105000, 54000, 21000]
maio = [88000, 52000, 22000]

resultado_firnanceiro_operacional_janeiro = janeiro[0] - janeiro[1] - janeiro[2]
janeiro.append(resultado_firnanceiro_operacional_janeiro)
resultado_firnanceiro_operacional_fevereiro = fevereiro[0] - fevereiro[1] - fevereiro[2]
fevereiro.append(resultado_firnanceiro_operacional_fevereiro)
resultado_firnanceiro_operacional_marco = marco[0] - marco[1] - marco[2]
marco.append(resultado_firnanceiro_operacional_marco)
resultado_firnanceiro_operacional_abril = abril[0] - abril[1] - abril[2]
abril.append(resultado_firnanceiro_operacional_abril)
resultado_firnanceiro_operacional_maio = maio[0] - maio[1] - maio[2]
maio.append(resultado_firnanceiro_operacional_maio)

indice_margem_operacional_janeiro = (resultado_firnanceiro_operacional_janeiro / janeiro[0]) * 100
janeiro.append(indice_margem_operacional_janeiro)
indice_margem_operacional_fevereiro = (resultado_firnanceiro_operacional_fevereiro / fevereiro[0]) * 100
fevereiro.append(indice_margem_operacional_fevereiro)
indice_margem_operacional_marco = (resultado_firnanceiro_operacional_marco / marco[0]) * 100
marco.append(indice_margem_operacional_marco)
indice_margem_operacional_abril = (resultado_firnanceiro_operacional_abril / abril[0]) * 100
abril.append(indice_margem_operacional_abril)
indice_margem_operacional_maio = (resultado_firnanceiro_operacional_maio / maio[0]) * 100
maio.append(indice_margem_operacional_maio)

IMOs = [indice_margem_operacional_janeiro, indice_margem_operacional_fevereiro, indice_margem_operacional_marco, indice_margem_operacional_abril, indice_margem_operacional_maio]
maior_IMO = max(IMOs)
mes_do_maior_IMO = IMOs.index(maior_IMO) + 1
menor_IMO = min(IMOs)
mes_do_menor_IMO = IMOs.index(menor_IMO) + 1

if IMOs[0] >= 25:
    janeiro.append("Excelente")
elif IMOs[0] >= 15:
    janeiro.append("Bom")
elif IMOs[0] >= 8:
    janeiro.append("Regular")
elif IMOs[0] > 0:
    janeiro.append("Atenção")
elif IMOs[0] == 0:
    janeiro.append("Ponto de equilibrio")
else:
    janeiro.append("Prejuízo")

if IMOs[1] >= 25:
    fevereiro.append("Excelente")
elif IMOs[1] >= 15:
    fevereiro.append("Bom")
elif IMOs[1] >= 8:
    fevereiro.append("Regular")
elif IMOs[1] > 0:
    fevereiro.append("Atenção")
elif IMOs[1] == 0:
    fevereiro.append("Ponto de equilibrio")
else:
    fevereiro.append("Prejuízo")

if IMOs[2] >= 25:
    marco.append("Excelente")
elif IMOs[2] >= 15:
    marco.append("Bom")
elif IMOs[2] >= 8:
    marco.append("Regular")
elif IMOs[2] > 0:
    marco.append("Atenção")
elif IMOs[2] == 0:
    marco.append("Ponto de equilibrio")
else:
    marco.append("Prejuízo")

if IMOs[3] >= 25:
    abril.append("Excelente")
elif IMOs[3] >= 15:
    abril.append("Bom")
elif IMOs[3] >= 8:
    abril.append("Regular")
elif IMOs[3] > 0:
    abril.append("Atenção")
elif IMOs[3] == 0:
    abril.append("Ponto de equilibrio")
else:
    abril.append("Prejuízo")

if IMOs[4] >= 25:
    maio.append("Excelente")
elif IMOs[4] >= 15:
    maio.append("Bom")
elif IMOs[4] >= 8:
    maio.append("Regular")
elif IMOs[4] > 0:
    maio.append("Atenção")
elif IMOs[4] == 0:
    maio.append("Ponto de equilibrio")
else:
    maio.append("Prejuízo")


print("Maior IMO:", maior_IMO)
print("Menor IMO:", menor_IMO)
print("Mês do maior IMO:", mes_do_maior_IMO)
print("Mês do menor IMO:", mes_do_menor_IMO)

print("Janeiro:", janeiro)
print("Fevereiro:", fevereiro)
print("Março:", marco)
print("Abril:", abril)
print("Maio:", maio)
```
