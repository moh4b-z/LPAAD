```python
def registrar_treinamento(tema, *participantes):
    print(f"O treinamento sobre {tema} foi registrado com os seguintes participantes:")
    for i, participante in enumerate(participantes, start=1):
        print(f"{i}. {participante}")
```

```python
registrar_treinamento("Python Avançado", "Julia", "Le", "Mohammad")
```

```python
def registrar_ocorrencia(codigo, *sistemas, **informacoes):
    print(f"Ocorrência registrada com código {codigo}.")
    print(f"Foram afetados {len(sistemas)} sistemas:")
    for i, sistema in enumerate(sistemas, start=1):
            print(f"{i}. {sistema}")

    if informacoes:
        print("Informações adicionais:")
        for chave, valor in informacoes.items():
            print(f"{chave}: {valor}")
```

```python
registrar_ocorrencia("INC-001", "Sistema da Casa da Leticia", "Sistema do SENAI", "API", prioridade="Alta", responsavel="Lapa")
```

```python
registrar_ocorrencia("INC-001")
```

```python
import statistics
```

```python
def vendas_dos_vendedores(vendedor, *vendas):
    if vendas:
        total_vendas = sum(vendas)
        media_vendas = statistics.mean(vendas)
        maior_venda = max(vendas)
    else:
        total_vendas = 0
        media_vendas = 0
        maior_venda = 0

    
    print(f"\n Vendedor: {vendedor}")
    print(f"Total de vendas: {total_vendas}")
    print(f"Média de vendas: {media_vendas}")
    print(f"Maior venda: {maior_venda}")
    if total_vendas > 5000:
        print(f"{vendedor} Você atingiu a meta de vendas.")
    else:
        print(f"{vendedor}, você não atingiu a meta de vendas.")
```

```python
vendas_dos_vendedores("José", 1000, 1500, 2000, 1200, 1800, 1200, 666, 999)
vendas_dos_vendedores("Pedro", 1200, 1800, 1200, 666)
vendas_dos_vendedores("Lucas", 800, 1300, 1700, 100)
```

```python
def registro_dos_parcientes(nome_paciente, *procedimentos, **informacoes):    
    print(f"\n O paciente {nome_paciente}")
    print(f"Passou por esse procedimentos:")
    print(procedimentos)
    print("Informações complementais:")
    print(informacoes)
    if "retorno" not in informacoes:
        print("Retorno não informado")
```

```python
registro_dos_parcientes(
    "Leticia", 
    "Consulta Geral", "Hemograma", "Raio-X", 
    idade=34, convenio="Saúde Top", retorno="11/09/2001"
)

registro_dos_parcientes(
    "Julia", 
    "Limpeza Dentária", 
    idade=22, prioridade="Alta"
)
registro_dos_parcientes(
    "Lipi", 
    telefone="11 93430 5345", cidade="São Paulo"
)
```

```python
def cadastrar_funcionario(nome, idade, salario, setor):
    if not nome:
        return "Nome é obrigatorio"
    if not isinstance(nome, str):
        return "Nome é obrigatorio"
    if 18 <= idade and idade <= 70:
        return "Idade inválida"
    if salario <= 0:
        return "Salario inválido"
    if setor in ["TI", "RH", "Finaceiro", "Comercial"]:
        return "Setor inválido"

    return "Funcionario cadastrado com sucesso"
```

```python
cadastrar_funcionario(2 ,2 ,2 ,2)
```
